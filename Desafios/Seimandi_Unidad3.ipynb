## Actividad 3. Validación y decisiones

### Procedimiento para validar un registro

```text
PROCEDIMIENTO procesar_registro(linea)

    Separar la linea usando "|" y contar los campos.
    Si no hay 5 campos, rechazar por cantidad incorrecta de campos.

    Quitar los espacios al inicio y al final de cada campo.
    Pasar el identificador a mayusculas.
    Pasar el correo y la categoria a minusculas.

    Si el identificador no es T seguida de tres digitos del 0 al 9,
        rechazar por identificador invalido.

    Si el nombre esta vacio,
        rechazar por nombre vacio.

    Si el correo no tiene exactamente una @ o contiene espacios,
        rechazar por correo invalido.
    Si el usuario del correo esta vacio, el dominio no tiene punto
    o hay un segmento vacio en el dominio,
        rechazar por correo invalido.

    Si la categoria no esta en el catalogo,
        rechazar por categoria invalida.

    Si el campo de minutos esta vacio,
        rechazar por minutos vacios.
    Si contiene algun caracter que no sea un digito del 0 al 9,
        rechazar por minutos no enteros o negativos.

    Convertir los minutos a entero.
    Si los minutos son mayores que 180,
        rechazar por minutos fuera de rango.

    Devolver los cinco campos limpios y un motivo vacio.

FIN PROCEDIMIENTO
```

Cuando una regla falla, se devuelve `(None, primer motivo)` y se termina la revisión. Al procesar el conjunto de registros, se guarda la línea original junto con su motivo de rechazo.

#### Diagrama para clasificar el tiempo

Una vez que los minutos son válidos, se comparan con el límite de 15.

```text
                        (Inicio)
                            |
                            v
                    / Leer minutos /
                            |
                            v
                            /\
                           /  \
                          /    \
                         /      \
      +------ Si -------< min<=15? >------- No ------+
      |                  \      /                    |
      v                   \    /                     v
 [En tiempo]               \  /              [Fuera de tiempo]
      |                     \/                       |
      +---------------------+------------------------+
                            |
                            v
                          (Fin)
```

- **14 minutos** → Sí → En tiempo.
- **15 minutos** → Sí → En tiempo. El límite está incluido.
- **16 minutos** → No → Fuera de tiempo. El registro es válido y supera el límite de atención.

````python
#Validoción tiempo.

limite = 15

for minutos in [14, 15, 16]:
    if minutos <= limite:
        print(minutos, "minutos: En tiempo")
    else:
        print(minutos, "minutos: Fuera de tiempo")
````

### Actividad 4. Funciones para procesar los registros

En esta actividad recorro la lista `registros`. En cada vuelta del `for`, la variable `linea` toma un registro completo y se lo pasa a `procesar_registro(linea)`. Cada registro viene como texto, con cinco campos separados por `|`.

#### Funciones que definí

| Función | ¿Qué recibe? | ¿Qué hace y qué devuelve? |
|---|---|---|
| `limpiar_texto(texto)` | El texto de un campo. | Uso `strip()` para quitar los espacios de los extremos, `split()` para separar las palabras y `" ".join(...)` para unirlas con un solo espacio. Devuelvo el texto limpio. |
| `correo_basico(correo)` | El correo extraído del registro. | Uso `count("@")` para comprobar que haya una sola arroba y `split("@")` para separar el usuario del dominio. Reviso que el usuario tenga texto, que el correo no tenga espacios y que el dominio tenga al menos un punto y ningún segmento vacío. Devuelvo `True` si pasa la revisión y `False` si falla. |
| `procesar_registro(linea)` | Una línea completa de la lista `registros`. | Uso `split("|")` para separar los campos y `len()` para comprobar que sean cinco. Limpio cada campo, paso el ticket a mayúsculas con `upper()` y el correo y la categoría a minúsculas con `lower()`. Después reviso el ticket, el nombre, el correo, la categoría y los minutos, en ese orden. |
| `clasificar_tiempo(minutos)` | Una cantidad de minutos ya validada. | Uso `minutos <= 15` para devolver `"en tiempo"` o `"fuera de tiempo"`. Por ejemplo, 15 queda en tiempo y 16 queda fuera. |

#### Cómo guardo los resultados

En `procesar_registro`, si una regla falla, uso `return` para terminar la revisión y devolver `None` junto con el primer motivo de rechazo. Si todos los campos son válidos, convierto los minutos a entero con `int()`, compruebo que estén dentro del rango permitido y devuelvo la fila limpia con un motivo vacío (`""`).

En el ciclo, `fila, motivo = procesar_registro(linea)` guarda cada resultado en su variable. Si `fila` es `None`, uso `append()` para agregar `[linea, motivo]` a `rechazados`. Si la fila es válida, la agrego a `aceptados`. Así puedo conservar el texto original de cada rechazo y saber qué regla falló.

El ciclo final separa los registros aceptados de los rechazados. La función `clasificar_tiempo` se puede probar aparte con 14, 15 y 16 minutos.

````python
# ACTIVIDAD 4. Validación de registros y clasificación del tiempo.

def limpiar_texto(texto):
    # Recibo el texto de un campo. Con strip() quito los espacios de los extremos.
    # Con split() separo las palabras y con join() las vuelvo a unir con un espacio.
    # Por ejemplo, "  Ana   López  " queda como "Ana López".
    return " ".join(texto.strip().split())


def correo_basico(correo):
    # Con count("@") cuento las arrobas. Con " " in correo busco un espacio.
    # Si cualquiera de las dos reglas falla, devuelvo False.
    if correo.count("@") != 1 or " " in correo:
        return False

    # Como ya comprobé que hay una sola arroba, split("@") da dos partes.
    # En la posición 0 queda el usuario y en la posición 1 queda el dominio.
    partes_correo = correo.split("@")
    usuario = partes_correo[0]
    dominio = partes_correo[1]

    # El usuario debe tener texto y el dominio debe tener al menos un punto.
    if usuario == "" or "." not in dominio:
        return False

    # Separo el dominio por puntos y reviso cada parte.
    # Por ejemplo, "ejemplo..org" se separa en ["ejemplo", "", "org"].
    # Esa parte vacía hace que el correo se rechace.
    for parte in dominio.split("."):
        if parte == "":
            return False

    # Si llego hasta aquí, el correo pasó todas las revisiones básicas.
    return True


# Esta lista contiene las categorías que puedo aceptar.
categorias = ["datos", "programacion", "visualizacion"]


def procesar_registro(linea):
    # El parámetro linea recibe el registro completo como texto.
    # Con split("|") lo separo y guardo las partes en campos_originales.
    # Se llaman originales porque todavía conservan sus espacios y mayúsculas.
    campos_originales = linea.split("|")

    # Con len() cuento los campos. Debe haber cinco.
    # Si faltan o sobran campos, devuelvo None para la fila y el motivo del rechazo.
    if len(campos_originales) != 5:
        return None, "numero de campos"

    # Empiezo con una lista vacía. El for toma un campo a la vez.
    # Limpio ese campo con limpiar_texto() y lo agrego a la lista con append().
    campos = []
    for campo in campos_originales:
        campos.append(limpiar_texto(campo))

    # Tomo cada dato por su posición. Las posiciones empiezan en 0.
    # Con upper() paso el ticket a mayúsculas y con lower() paso el correo
    # y la categoría a minúsculas. El nombre conserva sus mayúsculas y minúsculas.
    ticket = campos[0].upper()
    nombre = campos[1]
    correo = campos[2].lower()
    categoria = campos[3].lower()
    tiempo = campos[4]

    # El ticket debe tener cuatro caracteres, como T001.
    # ticket[0] me da el primer carácter, que debe ser T.
    if len(ticket) != 4:
        return None, "identificador"
    if ticket[0] != "T":
        return None, "identificador"

    # Con ticket[1:] tomo lo que viene después de la T.
    # Reviso que cada uno de esos tres caracteres sea un dígito del 0 al 9.
    for caracter in ticket[1:]:
        if caracter not in "0123456789":
            return None, "identificador"

    # Cada return termina la función en ese momento.
    # Por eso, si fallan varias reglas, solo devuelvo el primer motivo.
    if nombre == "":
        return None, "nombre vacio"

    # correo_basico() devuelve True o False. Con not entro al if si devuelve False.
    if not correo_basico(correo):
        return None, "correo"

    # Busco la categoría en la lista de categorías permitidas.
    if categoria not in categorias:
        return None, "categoria"

    # El tiempo todavía está guardado como texto.
    # Primero reviso si el campo quedó vacío después de limpiarlo.
    if tiempo == "":
        return None, "minutos vacios"

    # Reviso el tiempo carácter por carácter antes de convertirlo.
    # Por ejemplo, "-3" se rechaza porque el signo "-" no está entre los dígitos.
    for caracter in tiempo:
        if caracter not in "0123456789":
            return None, "minutos no enteros o negativos"

    # Como el texto ya pasó la revisión de dígitos, lo convierto a entero.
    # El máximo permitido es 180. El 180 se acepta porque uso > y no >=.
    minutos = int(tiempo)
    if minutos > 180:
        return None, "minutos fuera de rango"

    # Devuelvo la lista con los cinco datos y el motivo vacío.
    # En esta lista, minutos ya es un número entero.
    return [ticket, nombre, correo, categoria, minutos], ""


def clasificar_tiempo(minutos):
    # Recibo una cantidad de minutos que ya pasó la validación.
    # El <= incluye el 15. Por ejemplo, 15 queda en tiempo y 16 queda fuera.
    if minutos <= 15:
        return "en tiempo"
    return "fuera de tiempo"


aceptados = []
rechazados = []

# En cada vuelta, linea toma un registro de la lista registros.
# Le paso ese registro a la función para revisarlo.
for linea in registros:
    # Guardo el primer resultado en fila y el segundo en motivo.
    fila, motivo = procesar_registro(linea)

    if fila is None:
        # Si la función devolvió None, guardo el texto original y el motivo.
        rechazados.append([linea, motivo])
    else:
        # Si recibí una fila válida, la agrego a la lista de aceptados.
        aceptados.append(fila)

# Al final muestro cómo quedaron separados los registros.
print("ACEPTADOS:", aceptados)
print("RECHAZADOS:", rechazados)
````

## Actividad 5 · Indicadores y visualización.

Uso las listas de la actividad anterior para resumir cuántas solicitudes se aceptaron, cuánto tardaron y cómo se distribuyeron por categoría.

#### Total de registros y porcentaje de aceptación

Con `len()` cuento los elementos de `registros`, `aceptados` y `rechazados`. Para calcular el porcentaje de aceptación, divido la cantidad de aceptados entre el total de registros y multiplico por 100. Si el total es cero, muestro **sin datos** para evitar una división entre cero. Uso `:.2f` para mostrar el porcentaje con dos decimales.

#### Promedio y clasificación de los tiempos

Recorro `aceptados` y tomo los minutos de cada solicitud con `fila[4]`, que corresponde al quinto campo. Voy sumando esos minutos y, al final, divido la suma entre la cantidad de solicitudes aceptadas. Si ninguna fue aceptada, muestro **sin datos**.

También paso los minutos a `clasificar_tiempo()`. Si el tiempo es de **15 minutos o menos**, sumo uno al contador de solicitudes en tiempo. Si supera los 15 minutos, sumo uno al contador de solicitudes fuera de tiempo. Muestro el ticket de cada solicitud con `fila[0]` junto con su clasificación.

#### Conteo y gráfica por categoría

Para cada categoría, recorro las solicitudes aceptadas y comparo `fila[3]` con la categoría que estoy contando. Uso `append()` para guardar cada total en `conteos`. Los totales quedan en el mismo orden que las categorías, así que cada barra recibe el conteo que le corresponde.

Con `graficar_barras()` preparo la gráfica. Uso `plt.figure()` para crear la figura, `plt.bar()` para dibujar las barras y `plt.title()`, `plt.xlabel()` y `plt.ylabel()` para identificar qué se está mostrando. Ajusto el espacio con `plt.tight_layout()`, guardo la imagen como `solicitudes_por_categoria.png` con `plt.savefig()` y la muestro con `plt.show()`.

````python
# ACTIVIDAD 5. Indicadores y visualización.

# Con len() cuento los elementos de cada lista.
# registros tiene todas las solicitudes originales.
# aceptados y rechazados vienen de la validación anterior.
total = len(registros)
cantidad_aceptados = len(aceptados)

print("Total:", total)
print("Aceptados:", cantidad_aceptados)
print("Rechazados:", len(rechazados))

# Para calcular el porcentaje, divido los aceptados entre todos los registros
# y multiplico por 100. Reviso que el total sea mayor que cero antes de dividir.
if total > 0:
    porcentaje = cantidad_aceptados / total * 100

    # Con :.2f muestro el resultado con dos decimales.
    # También muestro la base, que es el número de registros revisados.
    print(f"Aceptación: {porcentaje:.2f}% (base: {total} registros)")
else:
    print("Aceptación: sin datos")


# Empiezo en cero porque todavía no he sumado minutos ni contado solicitudes.
suma_minutos = 0
en_tiempo = 0
fuera_tiempo = 0

# En cada vuelta, fila toma un registro de la lista aceptados.
for fila in aceptados:
    # fila[4] contiene los minutos. Es el quinto dato porque se cuenta desde 0.
    suma_minutos = suma_minutos + fila[4]

    # Paso los minutos a la función y guardo el resultado en estado.
    # Con fila[0] muestro el ticket junto con su clasificación.
    estado = clasificar_tiempo(fila[4])
    print(fila[0], estado)

    # Sumo uno al contador que corresponde al estado de esta solicitud.
    if estado == "en tiempo":
        en_tiempo = en_tiempo + 1
    else:
        fuera_tiempo = fuera_tiempo + 1

# Divido la suma de minutos entre los aceptados porque de esas filas tomé los tiempos.
if cantidad_aceptados > 0:
    promedio = suma_minutos / cantidad_aceptados
    print(f"Promedio: {promedio:.2f} min (base: {cantidad_aceptados} aceptados)")
else:
    # Sin solicitudes aceptadas, no hay tiempos con los que calcular el promedio.
    print("Promedio: sin datos")

print("En tiempo / fuera de tiempo:", en_tiempo, fuera_tiempo)


# Voy a guardar un conteo para cada categoría, en el mismo orden de la lista categorias.
conteos = []

for categoria in categorias:
    # Reinicio el contador cada vez que empiezo a revisar otra categoría.
    cantidad = 0

    # Recorro las filas aceptadas y comparo fila[3] con la categoría actual.
    # fila[3] es el cuarto dato de la fila y contiene la categoría.
    for fila in aceptados:
        if fila[3] == categoria:
            cantidad = cantidad + 1

    # Cuando termino de revisar las filas, agrego el total de esa categoría.
    conteos.append(cantidad)

print("Categorías:", categorias)
print("Conteos:", conteos)


def graficar_barras(etiquetas, valores, titulo, color):
    # etiquetas tiene los nombres de las categorías y valores tiene sus conteos.
    # La primera etiqueta va con el primer conteo, y así con las demás.
    plt.figure(figsize=(7, 4))
    plt.bar(etiquetas, valores, color=color)

    # Agrego el título de la gráfica y los nombres de los ejes.
    plt.title(titulo)
    plt.xlabel("Categoría")
    plt.ylabel("Solicitudes aceptadas")

    # Hago que el eje vertical empiece en cero y ajusto el espacio
    # para que se vean bien el título y las etiquetas.
    plt.ylim(bottom=0)
    plt.tight_layout()

    # Guardo la gráfica como imagen PNG y luego la muestro en el notebook.
    plt.savefig("solicitudes_por_categoria.png", dpi=150)
    plt.show()


# Paso las categorías como etiquetas y sus conteos como alturas de las barras.
graficar_barras(categorias, conteos, "Solicitudes por categoría", "steelblue")
````
